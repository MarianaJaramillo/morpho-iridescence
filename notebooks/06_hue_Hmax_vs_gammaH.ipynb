# 06 · Base colour vs colour change (Hmax vs γH)
Each point is a species (mean ± SD across sweeps). **Hmax** is the colour at normal geometry (span 0); **γH** is how fast the colour shifts with angle. The Spearman correlation tests whether species with a bluer or greener base colour also shift more. *M. eugenia*, *M. marcus* and *M. anaxibia* have no peak, so their Hmax is not comparable and only their γH is shown (right panel).

```python
%matplotlib inline
import warnings; warnings.filterwarnings("ignore")
from morpho_gruson.results import load_processed
from morpho_gruson import plots as P
P.style()
R = load_processed("../gruson/data/processed")
```

```python
fig, info = P.Hmax_vs_gammaH(R)
fig.savefig("../figures/Hmax_vs_gammaH.png", dpi=170, bbox_inches="tight")
print(f"Spearman rho = {info['spearman_rho']:.2f}, p = {info['spearman_p']:.2f}, n = {info['n']}")
```
