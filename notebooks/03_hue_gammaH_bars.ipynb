# 03 · Which hue is most stable, which most iridescent? (γH)
γH is the rate at which the peak wavelength shifts with viewing span. Larger γH = stronger colour change with angle. Error bars: SD across sweeps. Hatched bars (*eugenia*, *marcus*, *anaxibia*) use the spectral-edge position, so only γH — not Hmax — is comparable.

```python
%matplotlib inline
import warnings; warnings.filterwarnings("ignore")
from morpho_gruson.results import load_processed
from morpho_gruson import plots as P
P.style()
R = load_processed("../gruson/data/processed")
```

```python
fig = P.gammaH_bars(R)
fig.savefig("../figures/gammaH_bars.png", dpi=170)
```
