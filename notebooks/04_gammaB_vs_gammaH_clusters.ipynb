# 04 · Does brightness directionality relate to colour change? (γB vs γH, clusters)
Each point is a species (mean ± SD). Ward clustering on standardised (ln γB, γH). The Spearman correlation between γB and γH tests whether narrow-lobe species also change colour more.

```python
%matplotlib inline
import warnings; warnings.filterwarnings("ignore")
from morpho_gruson.results import load_processed
from morpho_gruson import plots as P
P.style()
R = load_processed("../gruson/data/processed")
```

```python
fig, info = P.gammaB_vs_gammaH(R, k=3)
fig.savefig("../figures/gammaB_vs_gammaH.png", dpi=170)
print("Silhouette by k:", {k: round(v, 3) for k, v in info["silhouette"].items()})
print(f"Spearman rho = {info['spearman_rho']:.2f}, p = {info['spearman_p']:.2f}")
for g, sp in info["groups"].items(): print(f"{g}: {', '.join(sp)}")
```
