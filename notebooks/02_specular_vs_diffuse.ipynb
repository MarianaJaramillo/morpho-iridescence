# 02 · Which species are most specular, which most diffuse?
Species means (± SD across sweeps) of **Bmax** against **γB**. The empirical relation Bmax = k/γB follows from the lobe having an approximately constant area (Bmax·γB ≈ const.): a narrow lobe is tall (specular), a wide lobe is dim (diffuse).

```python
%matplotlib inline
import warnings; warnings.filterwarnings("ignore")
from morpho_gruson.results import load_processed
from morpho_gruson import plots as P
P.style()
R = load_processed("../gruson/data/processed")
```

```python
fig, info = P.specular_vs_diffuse(R)
fig.savefig("../figures/Bmax_vs_gammaB.png", dpi=170)
print(f"Bmax = {info['k']:.2f} / gammaB   R² = {info['R2']:.2f}")
```

```python
s = R["summary_by_species"].assign(Bmax_over_gammaB=lambda d: d.Bmax_mean / d.gammaB_deg_mean)
s.sort_values("Bmax_over_gammaB", ascending=False)[["species","Bmax_mean","gammaB_deg_mean","Bmax_over_gammaB"]].round(2)
```
